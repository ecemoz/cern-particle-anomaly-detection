```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
file_path = "../data/raw/Dimuon_SingleMu.csv"
df = pd.read_csv(file_path)
df.head()
```

```python
df.info()
df.columns
df.describe()
```

```python
df.columns
```

```python
processed_path = "../data/processed/dimuon_processed.parquet"
df.to_parquet(processed_path, index=False)
processed_path
```

```python
feature_cols = [
    "pt1", "eta1", "phi1", "E1", "Q1",
    "pt2", "eta2", "phi2", "E2", "Q2",
    "M"
]

df_features = df[feature_cols].copy()
df_features.head()
```

```python
features_path = "../data/features/dimuon_features.parquet"
df_features.to_parquet(features_path, index=False)

features_path
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))
plt.hist(df["M"], bins=80, color="steelblue")
plt.xlabel("Dimuon invariant mass (GeV)")
plt.ylabel("Event count")
plt.title("Dimuon Mass Spectrum")
plt.show()
```

```python
df_features["event_type"] = "dimuon"
df_features["energy"] = 7  # 7 TeV
```

```python
df_features.head()
```

```python
base_path = "../data/features/spark"
partition_path = f"{base_path}/event_type=dimuon/energy=7"
partition_path
```

```python
import os
os.makedirs(partition_path, exist_ok=True)
```

```python
output_file = f"{partition_path}/part-0000.parquet"
df_features.to_parquet(output_file, index=False)
output_file
```
